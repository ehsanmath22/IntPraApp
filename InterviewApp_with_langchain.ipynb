```python
import os

import streamlit as st
from dotenv import load_dotenv
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser
```

```python
load_dotenv()

# Instantiate the model, pointing LangChain's OpenAI client at OpenRouter.
llm = ChatOpenAI(
    model="openai/gpt-4o-mini",
    temperature=0,
    max_tokens=None,
    max_retries=2,
    base_url="https://openrouter.ai/api/v1",
    api_key=os.getenv("OPENROUTER_API_KEY"),
)
```

```python
prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "You are an expert interview coach. When given a question or topic, provide a clear, structured, and practical answer to help the user prepare for job interviews. Include a strong sample answer where appropriate, key points to hit, and a short tip on delivery. Keep it focused and easy to read."),
        ("human", "Email:{question}"),
    ]
)
```

```python
chain=prompt|llm
```

```python
response = chain.invoke({'question':"who are you"})
```

```python
print(response.content)
```
